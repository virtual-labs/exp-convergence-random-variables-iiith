```python
import json
from IPython.display import display, Markdown

# Load the JSON file
with open('pretest.json', 'r') as f:
    data = json.load(f)

# Display each question in markdown format
for i, q in enumerate(data['questions'], 1):
    question_md = f"### Question {i}: \n{q['question']}\n\n"
    options_md = ""
    for opt in ['a', 'b', 'c', 'd']:
        prefix = "✅ **" if opt == q["correctAnswer"] else ""
        suffix = "**" if opt == q["correctAnswer"] else ""
        options_md += f"- {prefix}({opt}) {q['answers'][opt]}{suffix}\n"

    explanation = q["explanations"].get(q["correctAnswer"], "No explanation provided.")

    explanation_md = f"\n> **Explanation:** {explanation}\n"
    separator = "---\n"

    display(Markdown(question_md + options_md + explanation_md + separator))
```
